**Split Dataset**

```python
x = data_set.drop('AQI', axis = 1)
y = data_set['AQI']
```

**Standardization**

```python
scaler = StandardScaler()
scaler.fit(x)

standard_data = scaler.transform(x)
scaled_x = standard_data
x = scaled_x
```

```python
x = np.asarray(x)
y = np.asarray(y)
```

**Model Selection**

```python
reg_models = [
    LinearRegression(),
    RandomForestRegressor(random_state=0),
    SVR(kernel='linear'),
    KNeighborsRegressor()
]
```

```python
def best_regression_model():
    for model in reg_models:
        scores = cross_val_score(model, x, y, cv=5, scoring='r2')  # Use R2 score
        mean_score = scores.mean() * 100
        mean_score = round(mean_score, 2)

        print(f'R2 Cross Validation Scores of {model}: {scores}')
        print(f'Average R2 Score of {model}: {mean_score}%')
        print('---------------------------------------')
```

```python
best_regression_model()
```

```python
reg_models = [
    LinearRegression(),
    RandomForestRegressor(random_state=0),
    SVR(kernel='linear'),
    KNeighborsRegressor()
]
```

```python
regression_hyperparameters = {
    'linear_regression_hyperparameter': {},
    'random_forest_regressor_hyperparameter': {
        'n_estimators': [10, 20, 50, 100]
    },
    'svr_hyperparameter': {
        'kernel': ['linear', 'poly', 'rbf', 'sigmoid'],
        'C': [1, 5, 10, 20]
    },
    'knn_hyperparameter': {
        'n_neighbors': [3, 5, 10]
    }
}
```

```python
regression_hyperparameter_keys = list(regression_hyperparameters.keys())
```

```python
def regression_model_selection(models_list, model_hyperparameter):
    result = []
    for i, model in enumerate(models_list):
        key = regression_hyperparameter_keys[i]
        params = model_hyperparameter[key]

        print(f"Running GridSearchCV for: {model}")
        print(f"Hyperparameters: {params}\n")

        # If no params (e.g., LinearRegression), skip GridSearch
        if not params:
            model.fit(x, y)
            score = model.score(x, y)
            result.append({
                'model used': model,
                'highest score': score,
                'best hyper parameter': 'N/A'
            })
        else:
            regressor = GridSearchCV(model, params, cv=5, scoring='r2')
            regressor.fit(x, y)

            result.append({
                'model used': regressor.best_estimator_,
                'highest score': regressor.best_score_,
                'best hyper parameter': regressor.best_params_
            })

    result_dataframe = pd.DataFrame(result, columns=['model used', 'highest score', 'best hyper parameter'])
    return result_dataframe
```

```python
print(regression_model_selection(reg_models, regression_hyperparameters))
```

**Train, Test, Split**

```python
x_train, x_test, y_train, y_test = train_test_split(scaled_x,y, test_size = 0.2, random_state = 42)
```

```python
print(y_train)
```

**Model Evaluation**

```python
model = RandomForestRegressor(n_estimators=100, random_state=0)
model.fit(x_train, y_train)
```

```python
# Evaluate the Model on Training Data
x_predict = model.predict(x_train)
x_data_accuracy = r2_score(y_train, x_predict)
print(f'The accuracy score of training data is: {x_data_accuracy}')
```

```python
# Evaluate the Model on Test Data
x_test_predict = model.predict(x_test)
x_test_data_accuracy = r2_score(y_test, x_test_predict)
print(f'The accuracy score of test data is: {x_test_data_accuracy}')
```

**Build Prediction Model**

```python
input_data = (55.31,91.81,18.01,30.28,30.76,56.4,0.67,16.99,68.82,1.68,14.01,1.65)
# Convert the input_data into numpy array
input_data_array = np.asarray(input_data)

# Reshaped the input_data_array
reshaped_input_data_array = input_data_array.reshape(1,-1)

# Convert input data to standard data
std_data = scaler.transform(reshaped_input_data_array)

prediction = model.predict(std_data)
print(f'AQI: {prediction[0]}')

if prediction[0] <= 50:
      print("Good")
elif prediction[0] <= 100:
      print("Satisfactory")
elif prediction[0] <= 200:
      print("Moderate")
elif prediction[0] <= 300:
      print("Poor")
elif prediction[0] <= 400:
      print("Very Poor")
else:
      print("Severe")
```
